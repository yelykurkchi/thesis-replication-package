```python
import pandas as pd
```

# Preprocessing data from the survey

```python
raw_participants_data = pd.read_csv(
    "../../data/raw/Ethical Behavior Perception of SDC.csv",
    keep_default_na=False,
    na_values=[""],
)
```

```python
# Filter out rows where the answer for 'Would you accept the above terms and continue participating in the study?' is "No"
raw_participants_data = raw_participants_data[raw_participants_data['Would you accept the above terms and continue participating in the study?'] != "No"]

# Drop Timestamps
raw_participants_data = raw_participants_data.drop(columns=['Timestamp', 'Would you accept the above terms and continue participating in the study?'])
```

```python
# Drop confidential columns
final_processed_data = raw_participants_data.drop(columns=['Full Name', 'Email'])
```

```python
# Assigning ID to participants
final_processed_data = final_processed_data.copy()
N = len(final_processed_data)
width = max(2, len(str(N)))          
final_processed_data["participant_id"] = [f"P{idx:0{width}d}" for idx in range(1, N + 1)]
cols = ["participant_id"] + [c for c in final_processed_data.columns if c != "participant_id"]
final_processed_data = final_processed_data[cols]
final_processed_data = final_processed_data.set_index("participant_id", drop=False)
```

Save the dataframe as a CSV file

```python
final_processed_data.to_csv("../../data/processed/final_processed_data.csv", index=False)
```
